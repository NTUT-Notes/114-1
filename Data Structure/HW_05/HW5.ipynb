```python
from collections import deque
import re

# Node class 
class Hnode:
    def __init__(self, key, item):
        self.key = key
        self.item = item
        self.parent : Hnode|None = None
        self.right  : Hnode|None = None
        self.left   : Hnode|None = None

    def getKey(self):
        return self.key

    def getItem(self):
        return self.item

    def getRightChild(self):
        return self.right

    def getLeftChild(self):
        return self.left

    def getParent(self):
        return self.parent

    def hasRightChild(self):
        return (self.right!=None)

    def hasLeftChild(self):
        return (self.left!=None)

    def isRoot(self):
        return (self.parent==None)

    def isLeaf(self):
        return ((self.right==None)and(self.left==None))

    def isRightChild(self):
        return (self.parent.right == self)

    def isLeftChild(self):
        return (self.parent.left==self)

    def setParent(self, p):
        self.parent=p

    def setKey(self, key):
        self.key = key

    def setItem(self, item):
        self.item = item
        
    def addRightChild(self, hnode):
        self.right=hnode
        hnode.setParent(self)

    def addLeftChild(self, hnode):
        self.left=hnode
        hnode.setParent(self)

    def swapDataWith(self, hnode):
        tempKey = self.key
        tempItem = self.item
        
        self.key = hnode.key
        self.item = hnode.item

        hnode.key = tempKey
        hnode.item = tempItem

# Heap class 
class Heap:
    def __init__(self):
        self.root: Hnode | None = None
        self.last: Hnode | None = None
        self.size=0

    def isEmpty(self):
        return (self.size==0)

    def getSize(self):
        return self.size

    def getRoot(self):
        return self.root

    def getLast(self):
        return self.last

    def getHighestPriority(self):
        return self.root.getKey()

    def setRoot(self, hnode):
        self.root=hnode

    def setLast(self, hnode):
        self.last=hnode

#
# methods for heaps
#

    def __getMinChild__(self, current: 'Hnode'):
        left_child = current.getLeftChild()
        right_child = current.getRightChild()

        if right_child is None:
            return left_child

        if left_child is None:
            return right_child

        if left_child.getItem() < right_child.getItem():
            return left_child
        else:
            return right_child
    #
    # downward adjustment from current node
    #
    def downwardHeapify(self, current: 'Hnode'):
        if current.isLeaf():
            return
        
        item = current.getItem()

        minNode = self.__getMinChild__(current)

        if item > minNode.getItem():
            current.swapDataWith(minNode)
            self.downwardHeapify(minNode)

        return
    #
    # upward adjustment toward the root from the current node
    #
    def upwardHeapify(self, current: 'Hnode'):
        if current.isRoot():
            return
        
        parent = current.getParent()

        if parent.getKey() > current.getKey():
            current.swapDataWith(parent)
            return self.upwardHeapify(parent)

    #
    # The most important operation for a heap, remove_Min (extract_Min())
    #
    def removeMin(self):
        if self.last is None:
            return
        
        self.size -= 1
        
        if self.last.isRoot():
            self.root = None
            self.last = None
            return

        old_last = self.last
        parent_of_last = old_last.getParent()

        if old_last.isRightChild():
            self.last = parent_of_last.getLeftChild()
            parent_of_last.right = None
            return

        current = old_last

        while not current.isRoot() and not current.isRightChild():
            current = current.getParent()
        
        if current.isRoot():
            while current.hasRightChild():
                current = current.getRightChild()
        else:
            current = current.getParent().getLeftChild()
            
            while current.hasRightChild():
                current = current.getRightChild()

        self.last = current
        
        parent_of_last.left = None

    #
    # Another important operation for a heap, insertion() (add())
    #
    def Insert(self, hnode: 'Hnode'):
        hnode.parent = None
        hnode.left = None
        hnode.right = None

        self.size += 1

        if self.size == 1:
            self.root = hnode
            self.last = hnode
            return

        path = bin(self.size)[3:]
       
        parent = self.root
        for step in path[:-1]:
            if step == '0':
                parent = parent.getLeftChild()
            else:
                parent = parent.getRightChild()

        direction = path[-1]
        
        if direction == '0':
            parent.addLeftChild(hnode)
        else: 
            parent.addRightChild(hnode)


        self.last = hnode

        self.upwardHeapify(hnode)
    #
    # For management, print the heap in pre-order
    #
    def printHeapPreOrder(self, i: 'Hnode'):
        if i is None:
            return

        if i.isLeaf():
            print(f"Leaf [ {i.getKey()} {i.getItem()} ]")
        else:
            print(f"Node [ {i.getKey()} {i.getItem()} ]")

        self.printHeapPreOrder(i.getLeftChild())
        self.printHeapPreOrder(i.getRightChild())

# decimal to binary using yield
def d2by(x):
    while x>0 :
        x,r=divmod(x,2)
        yield r 

# function for starting the task
def HeapwithEntriesInserted():
    #
    # read the input information from the default input text file into an
    # entry list, entry_list
    #

    with open("inFile.txt", "r") as file:
        nodes = [Hnode(
            int(n.split(" ")[0]),
            n.split(" ")[-1].replace("\n", "")
        ) for n in file.readlines()]
    
    #
    # initiating a heap object h
    #
    h=Heap()

    #
    # Do something here to build the heap
    #
    for n in nodes:
        h.Insert(n)

    #---------------------Print as the example on the homework sheet --------
    #                       will be adapted with the input
    #
    print("Heap size=", h.getSize(), "The highest priority is ", h.getHighestPriority())
    print("pre-order traversal:")
    h.printHeapPreOrder(h.getRoot())

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("insert 35, resume")
    h.Insert(Hnode(35, "resume"))
    
    print("insert 15, second")
    h.Insert(Hnode(15, "second"))

    print("insert 20, fourth")
    h.Insert(Hnode(20, "fourth"))

    print("Heap size=", h.getSize(), "The highest priority is ", h.getHighestPriority())
    print("pre-order traversal:")
    h.printHeapPreOrder(h.getRoot())
    
    print("deleteMin")
    h.removeMin()

    print("insert 40, nineth")
    h.Insert(Hnode(40, "nineth"))

HeapwithEntriesInserted()
```

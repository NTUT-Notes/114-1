```python
def product_rec(n, m):
    if m <= 0:
        return 0

    return n + product_rec(n, m-1) 

print("The product of 4 and 7 is", product_rec(4, 7))
```

```python
def product_ite(n, m):
  sum = 0
  for i in range(m):
    sum += n
  return sum

print("The product of 4 and 7 is", product_ite(4, 7))
```

```python
def chebyshev_recursive(n):
  """Return Chebyshev polynomial T_n(x) (first kind) recursively."""
  if n==0: return [1]
  if n==1: return [1, 0]
  tn_minus_1 = chebyshev_recursive(n - 1)
  tn_minus_2 = chebyshev_recursive(n - 2)

  term = [c*2 for c in tn_minus_1] + [0]

  for i in range(len(tn_minus_2)):
    index = -1-i
    term[index] = term[index] - tn_minus_2[index]

  return term


if __name__ == "__main__":
    # demo: print first 10 polynomials
    for n in range(11):
         print(f"The coefficients of T_{n}(x) in list = {chebyshev_recursive(n)}")
```

```python
def chebyshev_memoreccu(n, m_table):
  """Return Chebyshev polynomial T_n(x) (first kind) recursively."""
  if m_table[n] != None:
    return m_table[n]
  
  if n == 0:
    return [1]

  if n == 1:
    return [1, 0]
      
  tn_minus_1 = chebyshev_memoreccu(n-1, m_table)
  tn_minus_2 = chebyshev_memoreccu(n-2, m_table)
      
  term = [c*2 for c in tn_minus_1] + [0]

  for i in range(len(tn_minus_2)):
    index = -1-i
    term[index] = term[index] - tn_minus_2[index] 

  m_table[n] = term
  return m_table[n]

if __name__ == "__main__":
    # demo: print first 10 polynomials
    for n in range(11):
      m_table=[None] * (n+1)
      print(f"The coefficients of T_{n}(x) in list = {chebyshev_memoreccu(n, m_table)}")
```

```python
def __chebyshev_addoutrec_helper__(n):
  if n == 0:
    # T_0, T_1
    return ( [1], [1, 0] )

  tn_minus_1, tn = __chebyshev_addoutrec_helper__(n-1)
  
  tn_plus_1 = [c*2 for c in tn] + [0]

  for i in range(len(tn_minus_1)):
    index = -1-i
    tn_plus_1[index] = tn_plus_1[index] - tn_minus_1[index] 

  return (tn, tn_plus_1)

def chebyshev_addoutrec(n):
  """Return Chebyshev polynomial T_n(x) (first kind) recursively."""

  return __chebyshev_addoutrec_helper__(n)[0]

if __name__ == "__main__":
    # demo: print first 10 polynomials
    for n in range(11):
      print(f"The coefficients of T_{n}(x) in list = {chebyshev_addoutrec(n)}")
```

```python
def chebyshev_iterative(n):
  """Return Chebyshev polynomial T_n(x) (first kind) recursively."""
  if n == 0:
    return [1]
  
  tn_minus_2 = [1]
  tn_minus_1 = [1, 0]

  for _ in range(2, n+1):
    tn = [c*2 for c in tn_minus_1] + [0]
    for j in range(len(tn_minus_2)):
      index = -1-j
      tn[index] = tn[index] - tn_minus_2[index]
    
    tn_minus_2 = tn_minus_1
    tn_minus_1 = tn

  return tn_minus_1

if __name__ == "__main__":
    # demo: print first 10 polynomials
    for n in range(11):
         print(f"The coefficients of T_{n}(x) in list = {chebyshev_iterative(n)}")
```
